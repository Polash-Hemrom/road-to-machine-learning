# Variables and Data Types

```python
name = "Polash"
age = 25
height = 5.6
is_student = True

print(name)
print(age)
print(height)
print(is_student)
```

# Data Types

```python
# Integers

x = 10
y = -5

# Floats (decimals)
pi = 3.14
temperature = -10.5

# Complex numbers
z = 3 + 4j

# type checking
print(type(x))
print(type(pi))
```

# String

```python
# Single or double quotes
firstname = 'Polash'
lastname = 'Hemrom'


# String operations
full_name = firstname + " "+ lastname
print(full_name)

# String methods
text = "Hello World"
print(text.upper())
print(text.lower())
print(text.replace("World", "Python"))
print(len(text))
```

# Boolean

```python
is_true = True
is_false = False

# Boolean operations
result = is_true and is_false   # False
result = is_true or is_false    # True
result = not is_true            # False
```

# Type Conversion

```python
x = "123"
y = int(x)
z = float(x)
w = str(123)

print(x)
print(y)
print(z)
print(type(w))

```

# Control Flow

```python
# If/Else Statements

age = 18

if age >= 18:
    print('You are an Adult')
elif age >= 13:
    print('You are teenager')
else:
    print('You are Child')
```

# Comparison Operators

```python
x = 10
y = 5

print(x > y)
print(x < y)
print(x == y)
print(x!= y)
print(x >= y)
print(x <= y)
```

# Loops

```python
# Iterate over a list
fruits = ['apple', 'banana', 'cherry']
for fruit in fruits:
    print(fruit)

# Using range
for i in range(5):
    print(i)


# With index
for index, fruit in enumerate(fruits):
    print(f"{index}: {fruit}")
```

# While  Loop

```python
count = 0
while count < 5:
    print(count)
    count += 1
```

# Loop Control

```python
# Break: exit loop

for i in range(10):
    if i == 5:
        break
    print(i)


# Continue: skip iteration
for i in range(5):
    if i == 2:
        continue
    print(i)
```

# Functions

## Defining Fucntions

```python
def greet(name):
    """This function greets a person"""
    return f"Hello, {name}!"
message = greet("Polash")
print(message)
```

## Function Parameters

```python
# Default Parameters
def power (base, exponent = 2):
    return base ** exponent


print(power(3))
print(power(3,3))


# Keyword arguments
def introduce (name, age, city):
    return f"{name} is {age} years old and lives in {city}."

print(introduce(age = 25, city = 'New York', name = "Polash"))
```
